```python
import pandas as pd
import tensorflow as tf
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense , Conv2D , Flatten
from keras.datasets import mnist
```

```python
(x_train , y_train) , (x_test , y_test) = mnist.load_data()
```

```python
model = Sequential()

model.add(Conv2D(30 , kernel_size= (5,5) , padding = 'valid' , activation = 'relu',input_shape = (28,28,1)))
model.add(Conv2D(30 , kernel_size= (5,5) , padding = 'valid' , activation = 'relu'))
model.add(Conv2D(30 , kernel_size= (5,5) , padding = 'valid' , activation = 'relu'))

model.add(Flatten())
model.add(Dense(100 , activation = 'relu'))
model.add(Dense(10 , activation = 'softmax'))
```

```python
model.summary()
```

```python
model = Sequential()

model.add(Conv2D(30 , kernel_size= (5,5) , padding = 'same' , activation = 'relu',input_shape = (28,28,1)))
model.add(Conv2D(30 , kernel_size= (5,5) , padding = 'same' , activation = 'relu'))
model.add(Conv2D(30 , kernel_size= (5,5) , padding = 'same' , activation = 'relu'))

model.add(Flatten())
model.add(Dense(100 , activation = 'relu'))
model.add(Dense(10 , activation = 'softmax'))
```
